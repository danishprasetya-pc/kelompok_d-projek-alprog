```python
def analisis_teks(teks):
    jumlah_karakter = len(teks)
    jumlah_kata = len(teks.split())
    return jumlah_karakter, jumlah_kata

teks_uji = "Untuk kode R, format tabelnya sama. Contoh baris kode R seperti teks <- readline. hati_hati hai"
karakter, kata = analisis_teks(teks_uji)
print(f"Jumlah Karakter: {karakter}")
print(f"Jumlah Kata: {kata}")
```

```python
def analisis_nip(nip):
    if len(nip) != 18:
        return "NIP tidak valid. Harus 18 digit!"

    # Slicing string berdasarkan standar NIP
    tahun_lahir = nip[:4]
    tahun_diangkat = nip[8:12]
    jk_kode = nip[14]
    jenis_kelamin = "Laki-laki" if jk_kode == '1' else "Perempuan"

    return f"Tahun Lahir: {tahun_lahir}, Diangkat: {tahun_diangkat}, Kelamin: {jenis_kelamin}"
```

```python
print(analisis_nip("198012142005011001"))
```
